```python
import numpy as np
import matplotlib.pyplot as plt


def norma_matriz(q, A):
    # Calcula a $q$-norma da matriz $A$ (só para $q$=1,2,$\infty$)
    m = len(A)
    if q == 1:
        return np.max(np.sum(np.abs(A), axis=0))
    if q == 2:
        return np.sqrt(np.max(np.linalg.eigvals(A.T @ A)))
    if q == 'inf':
        return np.max(np.sum(np.abs(A), axis=1))

### Matriz inversa das implementações 2 e 3
def A_implem_2_e_3_inv(m):
    h = 1 / (m+1)
    A_implem_2_e_3 = (np.diag(np.ones(m-1), k=-1) - 2*np.diag(np.ones(m)) + np.diag(np.ones(m-1), k=1)) / h**2
    return np.linalg.inv(A_implem_2_e_3)

### Matriz inversa das implementações 4 (CC de Neumann avançada)
def A_implem_4_avancada_inv(m):
    h = 1 / (m+1)
    A_implem_4_avancada = np.diag(np.ones(m), k=-1) - 2*np.diag(np.ones(m+1)) + np.diag(np.ones(m), k=1)
    A_implem_4_avancada[0, :] = 0
    A_implem_4_avancada[0, 0] = -h
    A_implem_4_avancada[0, 1] = h
    A_implem_4_avancada = A_implem_4_avancada / h**2
    return np.linalg.inv(A_implem_4_avancada)

### Matriz inversa das implementações 4 (CC de Neumann centrada)
def A_implem_4_centrada_inv(m):
    h = 1 / (m+1)
    A_implem_4_centrada = np.diag(np.ones(m), k=-1) - 2*np.diag(np.ones(m+1)) + np.diag(np.ones(m), k=1)
    A_implem_4_centrada[0, :] = 0
    A_implem_4_centrada[0, 0] = -1
    A_implem_4_centrada[0, 1] = 1
    A_implem_4_centrada = A_implem_4_centrada / h**2
    return np.linalg.inv(A_implem_4_centrada)
```

```python
metodos = [A_implem_2_e_3_inv, A_implem_4_avancada_inv, A_implem_4_centrada_inv]
pontos_internos = [2**3, 2**4, 2**5, 2**6, 2**7, 2**8, 2**9, 2**10, 2**11, 2**12]
qs = [2, 'inf']
normas_todas = []

for metodo in metodos:
    normas_q = []

    ### Cálculos a $q$-norma do método atual e armazena em normas_q
    for q in qs:
        normas = []
        for m in pontos_internos:
            A_inv = metodo(m)
            normas.append(norma_matriz(q, A_inv))
        normas_q.append(normas)

    normas_todas.append(normas_q)
```

```python
### Monta a tabela das normas das matrizes
normas_q_2_e_3 = normas_todas[0]
normas_q_avancada = normas_todas[1]
normas_q_centrada = normas_todas[2]

print('-'*95)
print(f" {'':<4} | {'Implementações 2 e 3':^27} | {'Diferença Avançada (CC)':^27} | {'Diferença Centrada (CC)':^27}")
print(f" {'m':<4} | {'2-norma':^13} {'inf-norma':^13} | {'2-norma':^13} {'inf-norma':^13} | {'2-norma':^13} {'inf-norma':^13}")
print('-'*95)

for i in range(len(pontos_internos)):
    m = pontos_internos[i]

    e1_2 = normas_q_2_e_3[0][i]
    e1_inf = normas_q_2_e_3[1][i]

    e2_2 = normas_q_avancada[0][i]
    e2_inf = normas_q_avancada[1][i]

    e3_2 = normas_q_centrada[0][i]
    e3_inf = normas_q_centrada[1][i]

    print(f" {m:<4} | {e1_2.real:^13.6f} {e1_inf:^13.6f} | {e2_2.real:^13.6f} {e2_inf:^13.6f} | {e3_2.real:^13.6f} {e3_inf:^13.6f}")

print('-'*95)
```
